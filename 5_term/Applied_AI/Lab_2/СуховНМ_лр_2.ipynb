### Проверяем нет ли тестовые данные пустых значений значений (Nan)

Построим функцию `def missing_value_checker` для проверки и подсчёта пропущеных значений в test_data. А также выведем тип данных этих значений.


```python
def missing_value_checker(data):
    list = []
    for feature, content in data.items():
        if data[feature].isnull().values.any():
           
            sum = data[feature].isna().sum()
            type = data[feature].dtype
            print (f'{feature}: {sum}, type: {type}')
           
            list.append(feature)
    print(list)
    print(len(list))
missing_value_checker(test_data)
```

Проверяем какие признаки в таблице можно оставить, а какие удалить. Если пропущенных значений слишком много, то удалим признак. Если их небольшое количество, то заполним `mean` или `median` для чисел, новая категория `missing` для строковых объектов.

В соответствии с этим:

– удалим ['Alley', 'FireplaceQu', 'PoolQC', 'Fence', 'MiscFeature'];

– заполним числовое отсутствующее значение значением `mean`;

– заполним строковое отсутствующее значение значением `missing`.

```python
test_edited = test_data.drop(['Alley','FireplaceQu','PoolQC', 'Fence', 'MiscFeature'], axis=1)
train_edited = train_data.drop(['Alley','FireplaceQu','PoolQC', 'Fence', 'MiscFeature'], axis=1)

def nan_filler(data):
    for label, content in data.items():
        if pd.api.types.is_numeric_dtype(content):
            data[label] = content.fillna(content.median())
        else:
            data[label] = content.astype("category").cat.as_ordered()
            data[label] = pd.Categorical(content).codes+1

nan_filler(test_edited)
nan_filler(train_edited)
```

## Моделирование

### Построение и обучение модели


Создайте последовательную модель нейронной сети с помощью фрэймворков тренировки нейронных сетей как: Torch или Tensorflow. 

```python
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import GridSearchCV
import tensorflow as tf
import numpy as np
import warnings
warnings.filterwarnings('ignore')
tf.random.set_seed(40)
```

```python
# Создаем обертку для использования с GridSearchCV
class KerasGS:
    def __init__(self, units1=64, units2=32, activation='linear', optimizer='adam', epochs=100, batch_size=32, loss='mse'):
        self.units1 = units1
        self.units2 = units2
        self.activation = activation
        self.optimizer = optimizer
        self.epochs = epochs
        self.batch_size = batch_size
        self.loss = loss
        
    def create_model(self):
        model = keras.Sequential([
            layers.Dense(self.units1, activation=self.activation, input_shape=(X_train.shape[1],)),
            layers.Dense(self.units2, activation=self.activation),
            layers.Dense(1, activation='linear')
        ])
        
        model.compile(
            optimizer=self.optimizer,
            loss=self.loss,
            metrics=['mae']
        )
        return model
        
    def get_params(self, deep=True):
        return {
            'units1': self.units1,
            'units2': self.units2,
            'activation': self.activation,
            'optimizer': self.optimizer,
            'epochs': self.epochs,
            'batch_size': self.batch_size,
            'loss': self.loss
        }
        
    def set_params(self, **params):
        for key, value in params.items():
            setattr(self, key, value)
        return self
        
    def fit(self, X, y):
        self.model_ = self.create_model()
        self.history_ = self.model_.fit(
            X, y,
            epochs=self.epochs,
            batch_size=self.batch_size,
            verbose=0
        )
        return self
        
    def predict(self, X):
        return self.model_.predict(X)
        
    def score(self, X, y):
        loss = self.model_.evaluate(X, y, verbose=0)[0]
        return -loss
```

```python
# Набор параметров
param_grid = {
    'units1': [64, 128, 256],
    'units2': [64, 128, 256], 
    'activation': ['linear'],
    'optimizer': ['adam', 'sgd'],
    'epochs': [200],
    'batch_size': [16, 32, 64],
    'loss': ['mse', 'mape', 'mae']
}
base_model = KerasGS()

grid = GridSearchCV(
    estimator=base_model,
    param_grid=param_grid,
    cv=2,
    scoring='neg_mean_squared_error',
    n_jobs=-1,
    verbose=1
)

# Обучаем
grid_result = grid.fit(X_train.values, y_train.values)
```

```python
print("Лучшие параметры:", grid_result.best_params_)
print("Лучший score:", grid_result.best_score_)
```

Скомпилируйте нейронную сеть, выбрав функцию потерь и оптимизатор соответственно.

```python
best_params = grid_result.best_params_
final_model = keras.Sequential([
    layers.Dense(best_params['units1'], activation=best_params['activation'], input_shape=(X_train.shape[1],)),
    layers.Dense(best_params['units2'], activation=best_params['activation']),
    layers.Dense(1, activation='linear')
])

final_model.compile(
    optimizer=best_params['optimizer'],
    loss='mse',
    metrics=['mae']
)
```

Обучите модель на обучающих данных `X_train` и `y_train` задав гиперпараметры вашей модели нейронной сети, например количество эпох (epochs), размер мини-выборки (batch_size) и другие.

```python
history = final_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=best_params['epochs'],
    batch_size=best_params['batch_size'],
    verbose=1
)
```

#### Анализ влияния количества эпох

```python
plt.figure(figsize=(8, 10))

plt.subplot(2, 1, 1)
plt.plot(history.history['loss'], label='Training Loss', linewidth=2)
plt.plot(history.history['val_loss'], label='Validation Loss', linewidth=2)
plt.title('MSE')
plt.xlabel('Эпоха')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(2, 1, 2)
plt.plot(history.history['mae'], label='Training MAE', linewidth=2)
plt.plot(history.history['val_mae'], label='Validation MAE', linewidth=2)
plt.title('MAE')
plt.xlabel('Эпоха')
plt.ylabel('MAE')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Эпоха представляет собой полный проход по всем обучающим данным, итерация соответствует одному шагу обновления весов при обработке одной мини-выборки. В контексте обучения нейронной сети эпоха включает в себя много итераций - количество итераций в одной эпохе равно общему числу примеров поделить на размер мини-выборки.

Функция активации - нужна для того,чтобы нейросеть могла аппроксимировать нелинейные зависимости. Проимеры: ReLu, сигмоида (для бинарной классификации), гиперболический тангенс и линейную активацию (для регрессионных задач). В данной модели использовалась линейная активация, что типично для регрессионных проблем.

MSE (средняя квадратичная ошибка) вычисляется как среднее квадратов разностей между предсказанными и истинными значениями. MAE (средняя абсолютная ошибка) рассчитывается как среднее абсолютных разностей. MSE сильнее штрафует большие ошибки и чувствительна к выбросам, в то время как MAE более устойчива к выбросам и линейно штрафует ошибки. MSE чаще применяется как дифференцируемая функция для оптимизации в процессе обучения.
