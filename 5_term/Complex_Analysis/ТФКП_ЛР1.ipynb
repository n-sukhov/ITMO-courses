# Задание 1 
## Множество Мандельброта
### Доказательства свойств
**Множество Мандельброта переходит само в себя при сопряжении. Иными словами, оно симметрично относительно вещественной оси**

_Доказательство_

◄
Рассмотрим последовательность для произвольного $c$ и комплексно сопряженного ему элемента $\overline{c}$, удовлетворяющих определению множества Мандельброта (впрочем, даже если брать те $c$, которые мн-ву не принадлежат, т.е. для которых последовательность неограничена, то данные рассуждения всё ещё будут справедливыми, так как они не будут влиять на ограниченность).


Назовём орбитой $O(c)$ множество точек, получающихся последовательным применением формулы $z_{n+1}=z_n^2+c$ к начальной точке $z_0=0$.


Для $c$ имеем:
$O(c)=\{0, c, c^2+c, c^4+2c^3+c^2+c,\dots\}$

Для $\overline{c}$ имеем:
$O(\overline{c})=\{0, \overline{c}, \overline{c}^2+\overline{c}, \overline{c}^4+2\overline{c}^3+\overline{c}^2+\overline{c},\dots\}$

Нетрудно заметить, что все элементы принадлежащие орбите будут являться многочленами с вещественными коэффициентами. Поскольку операция сопряжения перестановочна с арифметическими операциями над комплексными числами, $\overline{z_1z_2}=\overline{z_1}\cdot\overline{z_2}, \quad \overline{z_1+z_2}=\overline{z_1}+\overline{z_2}$, для такого многочлена $P(z)$ будет верно $P(\overline{z})=\overline{P(z)}$. Следовательно, $O(\overline{c})=\overline{O(c)}$.

Таким, образом, мы получили, что для всех $c$ орбита сопряженных им $\overline{c}$ будет симметричная орбите для $c\Rightarrow$ множество симметрично относительно вещественной оси.
►

**Если |c| > 2, то c не принадлежит множеству Мандельброта**

_Доказательство_

◄
Имеем последовательность
$$z_{n+1}=z_n^2+c,\quad z_0=0,\quad c\in\mathbb{C}.$$
Покажем, что при $|c|>2$ последовательность $|z_n|\rightarrow\infty$.
Для начала докажем следующее

**Утверждение (1):** Если для некоторого $n$ выполняется $|z_n|>\text{max}\{2, |c|\}$, то $\lim_{n\rightarrow\infty}{|z_n|}=\infty$.

◃
$\sqsupset|z_n|>2$ и $|z_n|>|c|$. Можно записать $|z_n|=2+\varepsilon,\quad \varepsilon>0,\varepsilon\in\mathbb{R}$. Рассмотрим последовательность $z_{n+1}=z_n^2+c$, по неравенству треугольника:
$|z_{n+1}|=|z_n^2+c|\geq|z_n^2|-|c|=|z_n|^2-|c|$. Поскольку $|z_n|>|c|,\quad -|z_n|<-|c|$. Тогда:

$$
|z_n|^2-|c|>|z_n|^2-|z_n|=|z_n|(|z_n|-1)=|z_n|(2+\varepsilon-1)=|z_n|(\varepsilon+1)\Rightarrow\\
\Rightarrow |z_{n+1}|=|z_n^2+c|>|z_n|(\varepsilon+1),\quad |z_{n+1}|>|z_n|(\varepsilon+1)
$$
В то же время, $\frac{|z_n|}{2}=\frac{2+\varepsilon}{2}=1+\frac{\varepsilon}{2}$. Тогда:

$$
|z_{n+1}|>|z_n|(\varepsilon+1)>|z_n|(1+\frac{\varepsilon}{2})=\frac{|z_n|^2}{2}
$$

Мы получили, что при $|z_{n+1}|>\frac{|z_n|^2}{2}\Rightarrow$ последовательность растёт не медленнее геометрической прогрессии. Следовательно, последовательность не ограничена, $\lim_{n\rightarrow\infty}{|z_n|}=\infty$.
▹

Теперь допустим, что $|c|>2,\quad |c|=2+\delta,\quad \delta>0$.

$z_0=0,\quad z_1=c.\quad$ Пока $|z_1|=|c|$, утверждение (1) не выполняется. Рассмотрим $|z_2|=|z_1^2+c|>|z_1|^2-|c|$ (в силу неравенства треугольника). $z_1=c\Rightarrow |z_2|>|c|(|c|-1)=|c|(2+\delta-1)=|c|(1+\delta)>|c|$.

Мы получили, что уже при $n=2$ выполняется $|z_n|>2,\quad |z_n|>|c|$. С этого момента мы можем воспользоваться утверждением (1) $\Rightarrow$ последовательность неограничена и не принадлежит множеству Мандельброта.
►

## Задание 2
# Визуализация  множества Мандельброта

Исследуем последовательность комплексных чисел, заданную следующим образом:
\begin{aligned}
\quad z_{n+1} = z_n^2 + c,\
\end{aligned}
где $c$ — это комплексное число, которое меняется по всей области интереса, и для каждого $c$ определяется, принадлежит ли оно множеству Мандельброта исходя из поведения итерационной последовательности.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from typing import Callable
from dataclasses import dataclass
```

```python
def mandelbrot(c, max_iter):
    z = 0 + 0j
    for n in range(max_iter):
        z = z*z + c
        if abs(z) > 2:
            return n
    return max_iter

def mandelbrot_set(xmin, xmax, ymin, ymax, width, height, max_iter):
    # Создание сетки комплексных чисел
    real = np.linspace(xmin, xmax, width)
    imag = np.linspace(ymin, ymax, height)
    result = np.zeros((height, width))

    for i, im in enumerate(imag):
        for j, re in enumerate(real):
            c = complex(re, im)
            m = mandelbrot(c, max_iter)
            result[i, j] = m
    return result

def plot_mandelbrot(data, max_iter, title, cmap='twilight_shifted'):
    plt.figure(figsize=(8, 6))
    # Создаем изображение с пороговым окрашиванием
    # Все точки с значением == max_iter — чёрные
    display_data = np.copy(data)
    display_data[display_data == max_iter] = 0  # присваиваем 0 для "внутренних" точек
    plt.imshow(display_data, extent=(xmin, xmax, ymin, ymax), cmap=cmap, interpolation='bilinear', origin='lower')
    plt.colorbar(label='Количество итераций до выхода')
    plt.title(title)
    plt.xlabel('Re')
    plt.ylabel('Im')
    plt.show()

# Параметры 1-й области (широкий вид)
xmin, xmax = -2.0, 0.5
ymin, ymax = -1.3, 1.3
width, height = 800, 600
max_iter = 150

mandelbrot_img_1 = mandelbrot_set(xmin, xmax, ymin, ymax, width, height, max_iter)
plot_mandelbrot(mandelbrot_img_1, max_iter, 'Множество Мандельброта. Область 1')

# Параметры 2-й области (увеличенный участок)
xmin, xmax = -0.75, -0.73
ymin, ymax = 0.14, 0.16
width, height = 800, 600
max_iter = 150

mandelbrot_img_2 = mandelbrot_set(xmin, xmax, ymin, ymax, width, height, max_iter)
plot_mandelbrot(mandelbrot_img_2, max_iter, 'Множество Мандельброта. Область 2 (увеличение)')
```

Количество итераций (max_iter) влияет на точность изображения: больше итераций позволяет выявить более тонкие детали и структуры множества, особенно при большом увеличении. Меньшее количество итераций даёт более грубое изображение, скрывая мелкие особенности. Поэтому для качественного построения множества важно выбрать достаточное число итераций.

## Задание 3  
# Визуализация множества Жюлиа  

В данном задании исследуется поведение последовательности комплексных чисел, определяемой формулой:
\begin{aligned}
\quad z_{n+1} = z_n^2 + c,\
\end{aligned}
где $c$ - фиксированное комплексное число, задающее форму множества Жюлиа.  
Каждая точка $z_0$ комплексной плоскости рассматривается как начальное значение, для которого вычисляется, покидает ли последовательность предел радиуса 2 при итерациях.

Для визуализации множества была реализована программа, выполняющая следующие шаги:

1. Создание равномерной сетки комплексных чисел в заданной области (определяется объектом *Viewport*).  
2. Итеративное вычисление функции для каждой точки до достижения максимального числа итераций или выхода за предел круга радиуса 2.  
3. Формирование двумерного массива результатов и визуализация при помощи цветовой карты.

В качестве параметра $c$ было выбрано значение:
\begin{aligned}
\quad c = -0.5251993 + 0.5251993i.
\end{aligned}

Были построены три изображения множества Жюлиа при разных масштабах области просмотра:

1. **Общий вид множества** — позволяет увидеть полную структуру фрактала. Количество итераций 200.
    
2. **Умеренное приближение** — детализирует отдельные участки узора. Количество итераций 300.
   
3. **Максимальное увеличение** — демонстрирует сложные самоподобные структуры на границе множества. Количество итераций 400.

   

Для визуализации использовалась цветовая карта `twilight_shifted`, позволяющая наглядно различать области с различным числом итераций до расходимости.

```python
# --- Класс для хранения параметров области отображения ---


@dataclass
class Viewport:
    xmin: float
    xmax: float
    ymin: float
    ymax: float
    width: int
    height: int
    max_iter: int


# --- Класс для вычисления и визуализации множества Жюлиа ---
class JuliaSet:
    """
    Класс выполняет расчёт значений множества Жюлиа
    и его визуализацию.
    """

    def __init__(self, c: complex):
        # Комплексный параметр c, определяет форму множества Жюлиа
        self.c = c

    @staticmethod
    def _iterate_point(z: complex, c: complex, max_iter: int) -> int:
        """
        Итерационная функция Жюлиа.
        Возвращает количество итераций до выхода из круга радиуса 2
        или максимальное число итераций, если выход не произошёл.
        """
        for n in range(max_iter):
            z = z * z + c
            if abs(z) > 2:
                return n
        return max_iter

    def compute(self, vp: Viewport) -> np.ndarray:
        """
        Вычисляет двумерный массив значений множества Жюлиа
        для заданного диапазона координат (Viewport).
        """
        # Создаём равномерную сетку точек в комплексной плоскости
        real = np.linspace(vp.xmin, vp.xmax, vp.width)
        imag = np.linspace(vp.ymin, vp.ymax, vp.height)
        result = np.zeros((vp.height, vp.width))

        # Проходим по каждой точке сетки и считаем количество итераций
        for i, im in enumerate(imag):
            for j, re in enumerate(real):
                z0 = complex(re, im)
                result[i, j] = self._iterate_point(z0, self.c, vp.max_iter)
        return result

    @staticmethod
    def plot(data: np.ndarray, vp: Viewport, cmap: str = 'twilight_shifted'):
        """
        Отрисовывает готовый массив множества Жюлиа как изображение.
        Цвета зависят от количества итераций до выхода точки за пределы круга радиуса 2.
        """
        plt.figure(figsize=(8, 6))

        # Копируем данные, чтобы не менять оригинал
        display_data = np.copy(data)
        # Точки, которые не вышли за пределы, делаем чёрными (значение = 0)
        display_data[display_data == vp.max_iter] = 0

        # Отрисовка изображения с нужными границами и цветовой картой
        plt.imshow(display_data, extent=(vp.xmin, vp.xmax, vp.ymin, vp.ymax),
                   cmap=cmap, interpolation='bilinear', origin='lower')
        plt.xlabel('Re(z)')
        plt.ylabel('Im(z)')


# --- Основная часть программы ---

# Комплексное число c, значение взято из задания
c = complex(-0.5251993, 0.5251993)

# Список значений основных параметров для трёх изображений с разным масштабом
viewports = [
    # 1. Общий вид множества (самое широкое окно)
    Viewport(xmin=-1.5, xmax=1.5, ymin=-1.2, ymax=1.2,
             width=600, height=900, max_iter=200),

    # 2. Увеличенная область (умеренный зум)
    Viewport(xmin=-0.4, xmax=-0.2, ymin=0.4, ymax=0.6,
             width=800, height=600, max_iter=300),

    # 3. Ещё более увеличенная область (максимальное приближение)
    Viewport(xmin=-0.351, xmax=-0.35, ymin=0.510, ymax=0.512,
             width=1000, height=900, max_iter=600),
]

# Создаём экземпляр множества Жюлиа с заданным c
julia = JuliaSet(c)

# Последовательно вычисляем и визуализируем каждую область
for vp in viewports:
    img = julia.compute(vp)
    julia.plot(img, vp)

# Отображаем все построенные изображения
plt.show()
```

**Вывод:**  
В ходе выполнения задания было реализовано построение множества Жюлиа для выбранного комплексного параметра. Полученные изображения подтвердили фрактальную природу множества: при увеличении масштаба наблюдаются новые повторяющиеся детали, демонстрирующие самоподобие и сложность структуры.

## Бассейны Ньютона
Фрактал Ньютона - это ограниченное множество на комплексной плоскости, которое описывется применением **метода Ньютона** к полиномиальной функции $p(z)\in\mathbb{C}[z]$ или трансцендентной функции. Фрактальные области появляются при нахождении приближённых корней нелинейного уравнения при помощи метода Ньютона на комплексной плоскости.

_Справка:_


Метод Ньютона - итерационный численный алгоритм нахождения нуля заданной функции по начальному приближению.Он заключается в последовательном применении следующей формулы до тех пор, пока не будет достигнута необходимая точность приближённого решения:
$$
z_{n+1}=z_n-\frac{f(z_n)}{f'(z_n)}
$$
Подробно останавливаться на этом не будем, но посмотрим на визуализация данного метода. Мы построили его так же в вещественной плоскости, потому что так более наглядно, визуализировать функцию из $f:\mathbb{C}\rightarrow\mathbb{C}$ не так удобно.

```python
def Newton_method(z_0: float, # Начальное приближение
        f: Callable[[float], float], # f(z)
        f_derivative: Callable[[float], float], # f'(z)
        max_iter: int, # Максимальное число итераций
        tol: float=1e-9 # Точность
        ):
    result = [z_0]
    for i in range(max_iter):
        if abs(f(result[i])) < tol:
            break
        result.append(result[i] -f(result[i]) / f_derivative(result[i])) 
        
    return result

#  Функции, на которых будем демонстрировать метод
def example_func(x):
    return 2 * x ** 3 + 6 * x ** 2 - 5 * x
def example_func_d(x):
    return 6 * x ** 2 + 12 * x - 5 
  
def example_complex_func(z):
    return z ** 3 - 1
def example_complex_func_d(z):
    return 3 * z ** 2

# Начальные приближения
x_0 = 2.0
z_0 = 3 + 2j

iters = Newton_method(x_0, example_func, example_func_d, 10)
iters_complex = Newton_method(z_0, example_complex_func, example_complex_func_d, 10)

fig, (ax_real, ax_complex) = plt.subplots(1, 2, figsize=(12, 6))

# Вещественная плоскость
x = np.linspace(-4, 4, 1000)
y = example_func(x)
ax_real.axhline(0, color='black', lw=1)
ax_real.axvline(0, color='black', lw=1)
ax_real.plot(x, y, 'blue', label='f(x)=2x^3+6x^2-5x')

# Находим корни
roots = np.roots([2, 6, -5, 0])
real_roots = roots[np.isreal(roots)].real
ax_real.plot(real_roots, np.zeros_like(real_roots), 'o', color='orange', markersize=8, label='Корни f(x) = 0')

# Подготовка объектов для анимации
point_real, = ax_real.plot([], [], 'ro', markersize=8)
tangent_line, = ax_real.plot([], [], 'r--', lw=1)
vertical_line, = ax_real.plot([], [], color="purple", linestyle="--", lw=1)

# Подписи к точкам
text_xn = ax_real.text(0, 0, '', color='red', fontsize=12, ha='center', va='bottom')
text_xn1 = ax_real.text(0, 0, '', color='darkred', fontsize=12, ha='center', va='top')

ax_real.set_title('Метод Ньютона, вещественный плоскость')
ax_real.set_xlabel('x')
ax_real.set_ylabel('y')
ax_real.set_xlim(-4, 4)
ax_real.set_ylim(-30, 50)
ax_real.grid(True, linestyle=':', alpha=0.7)

legend_xn = ax_real.text(0.02, 0.95, '', transform=ax_real.transAxes,
    fontsize=10, color='red', va='top')
legend_xn1 = ax_real.text(0.02, 0.90, '', transform=ax_real.transAxes,
    fontsize=10, color='darkred', va='top')
ax_real.legend()

# Комплексная плоскость
ax_complex.axhline(0, color='black', lw=1)
ax_complex.axvline(0, color='black', lw=1)

# Находим корни
roots_complex = np.roots([1, 0, 0, -1])
ax_complex.plot(roots_complex.real, roots_complex.imag, 'o', color='orange', markersize=8, label='Корни f(z)=0')

legend_zn = ax_complex.text(0.02, 0.95, '', transform=ax_complex.transAxes,
    fontsize=10, color='red', va='top')

ax_complex.set_xlim(-5, 5)
ax_complex.set_ylim(-5, 5)
ax_complex.set_title('Метод Ньютона, комплексная плоскость')
ax_complex.set_xlabel('Re(z)')
ax_complex.set_ylabel('Im(z)')
ax_complex.grid(True, linestyle=':', alpha=0.7)
legend_zn_current = ax_complex.text(0.02, 0.95, '', transform=ax_complex.transAxes,
    fontsize=10, color='red', va='top')
legend_zn_next = ax_complex.text(0.02, 0.90, '', transform=ax_complex.transAxes,
    fontsize=10, color='darkred', va='top')
ax_complex.legend(loc='upper right')

# Подготовка объектов для анимации
point_complex_current, = ax_complex.plot([], [], 'o', color='red', markersize=8)
point_complex_next, = ax_complex.plot([], [], 'o', color='darkred', markersize=8)
path_complex, = ax_complex.plot([], [], color='purple', linestyle='--', lw=1)
text_zn = ax_complex.text(0, 0, '', color='red', fontsize=12, ha='center', va='bottom')
text_zn1 = ax_complex.text(0, 0, '', color='darkred', fontsize=12, ha='center', va='bottom')

def update(frame):
    # Вещественная плоскость
    z = iters[frame]
    fx = example_func(z)
    fpx = example_func_d(z)

    # Касательная
    tangent_x = np.linspace(-4, 4, 2)
    tangent_y = fx + fpx * (tangent_x - z)
    point_real.set_data([z], [fx])
    tangent_line.set_data(tangent_x, tangent_y)
    vertical_line.set_data([z, z], [0, fx])

    # Подписи к точкам
    text_xn.set_position((z + 0.4, fx - 3))
    text_xn.set_text(f"$x_{{{frame}}}$")

    if frame + 1 < len(iters):
        x_next = iters[frame + 1]
        text_xn1.set_position((x_next, -2))
        text_xn1.set_text(f"$x_{{{frame+1}}}$")
        legend_xn.set_text(f"$x_{{{frame}}}$ = {z:.5f}")
        legend_xn1.set_text(f"$x_{{{frame+1}}}$ = {x_next:.5f}")
    else:
        text_xn1.set_text('')
        legend_xn.set_text(f"$x_{{{frame}}}$ = {z:.5f}")
        legend_xn1.set_text('')

    # Комплексная плоскость
    zc = iters_complex[frame]
    point_complex_current.set_data([zc.real], [zc.imag])
    text_zn.set_position((zc.real - 0.2, zc.imag + 0.2))
    text_zn.set_text(f"$z_{{{frame}}}$")

    # Путь метода
    re_vals = [p.real for p in iters_complex[:frame+1]]
    im_vals = [p.imag for p in iters_complex[:frame+1]]
    path_complex.set_data(re_vals, im_vals)

    if frame + 1 < len(iters_complex):
        z_next = iters_complex[frame + 1]
        point_complex_next.set_data([z_next.real], [z_next.imag])
        text_zn1.set_position((z_next.real - 0.2, z_next.imag - 0.5))
        text_zn1.set_text(f"$z_{{{frame+1}}}$")
        legend_zn_current.set_text(f"$z_{{{frame}}}$ = {zc.real:.5f} + {zc.imag:.5f}i")
        legend_zn_next.set_text(f"$z_{{{frame+1}}}$ = {z_next.real:.5f} + {z_next.imag:.5f}i")
    else:
        point_complex_next.set_data([], [])
        text_zn1.set_text('')
        legend_zn_current.set_text(f"$z_{{{frame}}}$ = {zc.real:.5f} + {zc.imag:.5f}i")
        legend_zn_next.set_text('')


    return (point_real, tangent_line, vertical_line, text_xn, text_xn1,
            legend_xn, legend_xn1,
            point_complex_current, point_complex_next, path_complex, text_zn, text_zn1)

ani = FuncAnimation(fig, update, frames=len(iters), interval=800, blit=True, repeat=False)

plt.tight_layout()
ani.save('newton_method_animation.gif', writer='pillow', fps=1)
```

##### Что же представляет из себя фрактал Ньютона?
По сути, данный объект есть ничто иное, как области, при выборе начальных точек для алгоритма Ньютона в которых, метод сойдется к тому или иному корню. Можно назвать их областями притяжения каждого из корней. Построим визуализацию. Оттенком цвета будем обозначать скорость сходимости к корню, темнее - более быстрая.

```python
def newton_fractal(coeffs: list, x_range: tuple, y_range: tuple,
        resolution: int=3000,
        max_iter: int=50, tol: float=1e-6, cmap: str='cool'):
    """   
    coeffs : Коэффициенты полинома
    x_range, y_range : Диапазоны значений по действительной и мнимой осям комплексной плоскости
    resolution : Разрешение сетки (количество точек вдоль каждой оси)
    max_iter : Максимальное количество итераций для метода Ньютона
    tol : Точность поиска корня
    cmap : цветовая палитра
    """
    # Построение бассейна Ньютона по заданным коэффициентам полинома

    # Создаём сетку точек на комплексной плоскости для расчёта сходимостей из них
    x = np.linspace(x_range[0], x_range[1], resolution)
    y = np.linspace(y_range[0], y_range[1], resolution)
    X, Y = np.meshgrid(x, y)
    Z = X + 1j * Y

    # Нахожим корни полинома
    roots = np.roots(coeffs)

    # Определяем сам полином и его производную
    def f(z): return np.polyval(coeffs, z)
    def f_der(z): return np.polyval(np.polyder(coeffs), z)

    # Карта индексов корней, к которым сошёлся каждый пиксель
    index_map = np.zeros(Z.shape, dtype=int)
    # Массив количества итераций для каждой точки
    iter_count = np.zeros(Z.shape)

    # Метод Ньютона
    for k in range(max_iter):
        not_converged = np.abs(f(Z)) > tol
        Z[not_converged] -= f(Z[not_converged]) / f_der(Z[not_converged])
        iter_count[not_converged] = k + 1

    # Определяем, к какому корню сошлась каждая точка
    for i, r in enumerate(roots):
        mask = np.abs(Z - r) < tol
        index_map[mask] = i

    # Получаем цветовую палитру, где каждому корню соответствует свой цвет
    cmap_obj = plt.get_cmap(cmap, len(roots))
    base_colors = cmap_obj(index_map / len(roots))

    # Добавляем яркость в зависимости от скорости сходимости
    brightness = iter_count / iter_count.max()
    fractal_rgb = base_colors.copy()
    fractal_rgb[..., :3] = fractal_rgb[..., :3] * (0.5 + 0.7 * brightness[..., np.newaxis])

    return fractal_rgb, roots

def newton_fractal_img(fractal_rgb: np.ndarray, roots: np.ndarray,
        filename: str = "Newton_fractal.png", show_roots: bool = True,
        x_range: tuple = (-2, 2), y_range: tuple = (-2, 2)):
    """
    fractal_rgb: Массив RGB, полученный из функции newton_fractal
    roots: корни
    x_range, y_range : Диапазоны значений по действительной и мнимой осям комплексной плоскости
    filename : Имя файла для сохранения результата
    show_roots : Отображать найденные корни
    """
    fig, ax = plt.subplots(figsize=(6, 6), facecolor="white")
    ax.set_facecolor('#1D1D96')
    ax.imshow(fractal_rgb, extent=(x_range[0], x_range[1], y_range[0], y_range[1]), alpha=0.9)

    if show_roots: # Отрисовка корней
       ax.scatter(roots.real, roots.imag, color='yellow', s=20, alpha=1.0, label="Корни")

    ax.set_title(f"Фрактал Ньютона для полинома степени {len(roots)}", color='black')
    ax.set_xlabel("Re(z)", color='black')
    ax.set_ylabel("Im(z)", color='black')
    ax.tick_params(axis='x', colors='black')
    ax.tick_params(axis='y', colors='black')

    ax.grid(False)
    ax.legend(facecolor='white')

    plt.savefig(filename, dpi=400)

fractal_rgb, roots = newton_fractal([1, 0, -1], x_range=(-2, 2), y_range=(-2, 2))
newton_fractal_img(fractal_rgb, roots, filename="Newton_fractal_2.png", x_range=(-2, 2), y_range=(-2, 2))

fractal_rgb, roots = newton_fractal([1, 0, 0, -1], x_range=(-2, 2), y_range=(-2, 2))
newton_fractal_img(fractal_rgb, roots, filename="Newton_fractal_3.png", x_range=(-2, 2), y_range=(-2, 2))

fractal_rgb, roots = newton_fractal([1, 0, 0, -1], x_range=(-0.1, 0.12), y_range=(0.25, 0.4))
newton_fractal_img(fractal_rgb, roots, filename="Newton_fractal_3_1.png", x_range=(-0.1, 0.12), y_range=(0.25, 0.4),
    show_roots=False)

fractal_rgb, roots = newton_fractal([1, 0, 0, -1], x_range=(0.025, 0.04), y_range=(0.31, 0.33))
newton_fractal_img(fractal_rgb, roots, filename="Newton_fractal_3_2.png", x_range=(0.025, 0.04), y_range=(0.31, 0.33),
    show_roots=False)
    
fractal_rgb, roots = newton_fractal([1, 0, 0, 0, -1], x_range=(-2, 2), y_range=(-2, 2))
newton_fractal_img(fractal_rgb, roots, filename="Newton_fractal_4.png", x_range=(-2, 2), y_range=(-2, 2))
```

```python
# Зададим промежуточный коэффициенты полиномом для вычисления корней
poly_list = [
    [1, 0, 0, -1],
    [2, 0, 1, -1],
    [3, -1, 2, -1],
    [1, -3, 4, -1],
]

fig, ax = plt.subplots(figsize=(6, 6), facecolor='white')
ax.set_facecolor('#0a0a3d')

x_range, y_range = (-2, 2), (-2, 2)
im = ax.imshow(np.zeros((400, 400, 3)), extent=(x_range[0], x_range[1], y_range[0], y_range[1]))
sc = ax.scatter([], [], color='yellow', s=30, alpha=1.0, label="Корни")

ax.set_title("Фрактал Ньютона", color='black')
ax.set_xlabel("Re(z)", color='black')
ax.set_ylabel("Im(z)", color='black')
ax.tick_params(axis='x', colors='black')
ax.tick_params(axis='y', colors='black')
ax.legend(facecolor='white')

def update(frame):
    # Определяем, между какими двумя полиномами мы сейчас интерполируем
    i = frame // 20
    t = (frame % 20) / 20

    # На последнем переходе останавливаемся, чтобы не выйти за границы списка
    if i >= len(poly_list) - 1:
        i = len(poly_list) - 2
        t = 1

    # Интерполируем коэффициенты
    coeffs = (1 - t) * np.array(poly_list[i]) + t * np.array(poly_list[i + 1])
    coeffs = coeffs.tolist()

    fractal_rgb, roots = newton_fractal(coeffs, x_range, y_range, resolution=1000, cmap='cool')

    im.set_data(fractal_rgb)
    sc.set_offsets(np.column_stack([roots.real, roots.imag]))
    return im, sc

ani = FuncAnimation(fig, update, frames=80, interval=200, blit=False, repeat=True)
ani.save("newton_fractal_animation.gif", fps=10, dpi=200)
plt.show()
```

Интересный факт заключается в том, что сколь малый круг мы бы ни выбрали на данном графике, в него попадет либо только 1 область притяжения, либо $n$ областей для каждого из корней полиномов.
