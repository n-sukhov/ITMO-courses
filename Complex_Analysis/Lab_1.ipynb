# Множество Мандельброта
### Доказательства свойств
**Множество Мандельброта переходит само в себя при сопряжении. Иными словами, оно симметрично относительно вещественной оси**

_Доказательство_

◄
Рассмотрим последовательность для произвольного $c$ и комплексно сопряженного ему элемента $\overline{c}$, удовлетворяющих определению множества Мандельброта (впрочем, даже если брать те $c$, которые мн-ву не принадлежат, т.е. для которых последовательность неограничена, то данные рассуждения всё ещё будут справедливыми, так как они не будут влиять на ограниченность).


Назовём орбитой $O(c)$ множество точек, получающихся последовательным применением формулы $z_{n+1}=z_n^2+c$ к начальной точке $z_0=0$.


Для $c$ имеем:
$O(c)=\{0, c, c^2+c, c^4+2c^3+c^2+c,\dots\}$

Для $\overline{c}$ имеем:
$O(\overline{c})=\{0, \overline{c}, \overline{c}^2+\overline{c}, \overline{c}^4+2\overline{c}^3+\overline{c}^2+\overline{c},\dots\}$

Нетрудно заметить, что все элементы принадлежащие орбите будут являться многочленами с вещественными коэффициентами. Поскольку операция сопряжения перестановочна с арифметическими операциями над комплексными числами, $\overline{z_1z_2}=\overline{z_1}\cdot\overline{z_2}, \quad \overline{z_1+z_2}=\overline{z_1}+\overline{z_2}$, для такого многочлена $P(z)$ будет верно $P(\overline{z})=\overline{P(z)}$. Следовательно, $O(\overline{c})=\overline{O(c)}$.

Таким, образом, мы получили, что для всех $c$ орбита сопряженных им $\overline{c}$ будет симметричная орбите для $c\Rightarrow$ множество симметрично относительно вещественной оси.
►

**Если |c| > 2, то c не принадлежит множеству Мандельброта**

_Доказательство_

◄
Имеем последовательность
$$z_{n+1}=z_n^2+c,\quad z_0=0,\quad c\in\mathbb{C}.$$
Покажем, что при $|c|>2$ последовательность $|z_n|\rightarrow\infty$.
Для начала докажем следующее

**Утверждение (1):** Если для некоторого $n$ выполняется $|z_n|>\text{max}\{2, |c|\}$, то $\lim_{n\rightarrow\infty}{|z_n|}=\infty$.

◃
$\sqsupset|z_n|>2$ и $|z_n|>|c|$. Можно записать $|z_n|=2+\varepsilon,\quad \varepsilon>0,\varepsilon\in\mathbb{R}$. Рассмотрим последовательность $z_{n+1}=z_n^2+c$, по неравенству треугольника:
$|z_{n+1}|=|z_n^2+c|\geq|z_n^2|-|c|=|z_n|^2-|c|$. Поскольку $|z_n|>|c|,\quad -|z_n|<-|c|$. Тогда:

$$
|z_n|^2-|c|>|z_n|^2-|z_n|=|z_n|(|z_n|-1)=|z_n|(2+\varepsilon-1)=|z_n|(\varepsilon+1)\Rightarrow\\
\Rightarrow |z_{n+1}|=|z_n^2+c|>|z_n|(\varepsilon+1),\quad |z_{n+1}|>|z_n|(\varepsilon+1)
$$
В то же время, $\frac{|z_n|}{2}=\frac{2+\varepsilon}{2}=1+\frac{\varepsilon}{2}$. Тогда:

$$
|z_{n+1}|>|z_n|(\varepsilon+1)>|z_n|(1+\frac{\varepsilon}{2})=\frac{|z_n|^2}{2}
$$

Мы получили, что при $|z_{n+1}|>\frac{|z_n|^2}{2}\Rightarrow$ последовательность растёт не медленнее геометрической прогрессии. Следовательно, последовательность не ограничена, $\lim_{n\rightarrow\infty}{|z_n|}=\infty$.
▹

Теперь допустим, что $|c|>2,\quad |c|=2+\delta,\quad \delta>0$.

$z_0=0,\quad z_1=c.\quad$ Пока $|z_1|=|c|$, утверждение (1) не выполняется. Рассмотрим $|z_2|=|z_1^2+c|>|z_1|^2-|c|$ (в силу неравенства треугольника). $z_1=c\Rightarrow |z_2|>|c|(|c|-1)=|c|(2+\delta-1)=|c|(1+\delta)>|c|$.

Мы получили, что уже при $n=2$ выполняется $|z_n|>2,\quad |z_n|>|c|$. С этого момента мы можем воспользоваться утверждением (1) $\Rightarrow$ последовательность неограничена и не принадлежит множеству Мандельброта.
►
