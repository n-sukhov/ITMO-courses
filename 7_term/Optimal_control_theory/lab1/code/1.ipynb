```python
import numpy as np
from typing import Callable

def grad_descent(
    z_0: np.ndarray,  # Начальное приближение
    grad_f: Callable[[np.ndarray], np.ndarray],  # Градиент функции
    max_iter: int,  # Максимальное число итераций
    gamma: float,  # Шаг
    tol: float = 1e-9,  # Точность по норме градиента
) -> np.ndarray:
    z = np.asarray(z_0, dtype=float).copy()
    result = [z.copy()]

    for _ in range(max_iter):
        grad = np.asarray(grad_f(z), dtype=float)

        if np.linalg.norm(grad) < tol:
            break

        z = z - gamma * grad
        result.append(z.copy())

    return np.array(result)


def J(x: float, u: float):
    return 2*x**2 + 3*u**2 + 4*x*u + 8*x + u - 4


def grad_J(z: np.ndarray) -> np.ndarray:
    x, u = z

    return np.array([
        4*x + 4*u + 8,
        4*x + 6*u + 1,
    ], dtype=float)


z_0 = np.array([0.0, 0.0])

for gamma in (0.1, 0.21):
    result = grad_descent(
        z_0=z_0,
        grad_f=grad_J,
        max_iter=1000,
        gamma=gamma,
    )

    x_min, u_min = result[-1]

    print(f"gamma = {gamma}")
    print(f"Число итераций: {len(result) - 1}")
    print(f"x = {x_min:.9f}, u = {u_min:.9f}")
    print(f"J = {J(x_min, u_min):.9f}")
```
