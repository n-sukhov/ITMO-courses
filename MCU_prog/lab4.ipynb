```python
import serial, matplotlib.pyplot as plt
ser = serial.Serial('COM10', 1000000, timeout=2)
data = []
for i in range(300):
    b = ser.read(2)
    if len(b) == 2:
        val = (b[0] << 8) | b[1]
        data.append(val * 3.3 / 4095)
ser.close()
plt.plot(data)
plt.title('Синусоида с АЦП')
plt.xlabel('Отсчеты')
plt.ylabel('Вольты')
plt.grid()
plt.show()
```
